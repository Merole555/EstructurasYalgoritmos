## Parte 2 · Inserción, y la lista casi ordenada (10 min)

Inserción funciona como ordenar cartas en la mano: toma el siguiente elemento y lo **desliza hacia
atrás** hasta que encuentre su lugar. Todo lo que está a su izquierda ya quedó ordenado.

La clave está en que **para de deslizar apenas encuentra un elemento menor** — o sea, se entera de que
ya está en su sitio. Guarden ese detalle: es lo que se van a medir en la celda de abajo.


```python
def insercion(lista):
    """
    Ordena la lista de menor a mayor con el método de inserción.
    """
    for i in range(1, len(lista)):
        actual = lista[i]      # la "carta" que estamos ubicando
        j = i - 1              # empezamos a mirar hacia atrás

        # TODO — te toca a ti.
        # Mientras j sea válido (j >= 0) Y lista[j] sea mayor que "actual":
        #     corre lista[j] una posición a la derecha  ->  lista[j + 1] = lista[j]
        #     y retrocede  ->  j -= 1
        #
        # Cuando el while termine, "actual" va en la posición j + 1.
        #
        # Ojo: las DOS condiciones del while van juntas con "and". Si te falta la de j >= 0,
        # el índice se sale por la izquierda y sobreescribes el final de la lista.

        lista[j + 1] = actual

    return lista


probar(insercion, "insercion")
```

## Parte 3 · Merge sort (15 min)

Aquí cambia la estrategia: **divide y vencerás**.

1. Si la lista tiene 0 o 1 elementos, ya está ordenada — no hay nada que hacer. *(caso base)*
2. Si no, se parte por la mitad, se ordena cada mitad **con la misma función**, y se mezclan las dos
   mitades ya ordenadas.

La función que mezcla dos listas ordenadas **ya está construida** más abajo: léanla, es el corazón del
algoritmo. Ustedes solo completan la parte recursiva.


```python
def mezclar(izquierda, derecha):
    """
    YA CONSTRUIDA — no hay que modificar nada aquí.

    Recibe dos listas YA ORDENADAS y devuelve una sola lista ordenada con todos los elementos.
    Solo mira el primer elemento pendiente de cada lista y se lleva el menor. Nunca retrocede:
    por eso mezclar cuesta O(n), un solo recorrido.
    """
    resultado = []
    i = j = 0

    while i < len(izquierda) and j < len(derecha):
        if izquierda[i] <= derecha[j]:      # <= (y no <) es lo que hace ESTABLE a merge sort
            resultado.append(izquierda[i])
            i += 1
        else:
            resultado.append(derecha[j])
            j += 1

    # una de las dos se acabó; lo que sobre de la otra ya está ordenado y va tal cual al final
    resultado.extend(izquierda[i:])
    resultado.extend(derecha[j:])
    return resultado


# Probemos mezclar por separado, antes de usarla:
print(mezclar([1, 4, 9], [2, 3, 10]))   # esperado: [1, 2, 3, 4, 9, 10]
```
